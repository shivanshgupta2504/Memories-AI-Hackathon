## Get the API Key

```python
API_KEY = os.getenv("MEMORIES_AI_API_KEY")
```

## Upload the Static file and get the `videoNo`

```python
headers = {"Authorization": API_KEY}  # API key
# file_path = r"C:\Users\Shivansh Gupta\OneDrive\Desktop\My Projects\AI-Projects\Memories-AI\ZXKH15nDQBsi.hevc.mp4"
file_path = r"C:\Users\Shivansh Gupta\OneDrive\Desktop\My Projects\AI-Projects\Memories-AI\Cat_Skateboarding_Video_Generated.mp4"
# Video file details
files = {
    "file": (os.path.basename(file_path), open(file_path, 'rb'), "video/mp4")
}

# Optional callback URL for task status notifications
data = {"unique_id": "123456"} # Optional
# If you are giving the Unique ID - It is like a folder under which you can upload many videos
# You have to use this ID everytime

response = requests.post(
    "https://api.memories.ai/serve/api/v1/upload",
    files=files,
    data=data,
    headers=headers
)

# print(response.json())

video_no = response.json().get("data", {}).get("videoNo", "")
video_no
```

## Check video parsing status

```python
headers = {"Authorization": API_KEY}
params = {
    "video_no": video_no, # Replace with the actual video number from the upload response, - works like filter
    "unique_id": "123456" # This is the folder ID under which we are uploading files
}
response = requests.post("https://api.memories.ai/serve/api/v1/list_videos", headers=headers, json=params)
print(response.json())

# Parsing takes time so you have to check continuously untill 'status': 'PARSE'
```

## Chat with Video - Non-streaming Mode

```python
import json

headers = {
    "Authorization": API_KEY,
    "Content-Type": "application/json",
}

payload = {
    "video_nos": [video_no],  # List of video IDs to chat about
    "prompt": "What is the Cat doing in this video?",  # User query
    "session_id": "123",  # Chat session ID
    "unique_id": "123456",
}

response = requests.post(
    "https://api.memories.ai/serve/api/v1/chat",
    headers=headers,
    json=payload,
    stream=False
)

content = response.json().get("data", {}).get("content", "")

# if response.status_code != 200:
#     print(response.status_code)
#     print(response.text)
# else:
#     try:
#         for line in response.iter_lines(decode_unicode=True):
#             if line:
#                 print(line)
#                 if line.strip().lower() == 'data:"done"':
#                     print("\n")
#                     break
#                 if line.startswith("data:"):
#                     print(line.replace("data:", "").strip(), end="", flush=True)
#     except Exception as e:
#         print(str(e))
```
