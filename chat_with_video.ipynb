## Video Captioning

```python
import json

url = "https://security.memories.ai/v1/understand/uploadFile"
headers = {"Authorization": API_KEY}

# JSON request body
data = {
    "user_prompt": "Summarize the video content and identify key persons",
    "system_prompt": "You are a video understanding system that summarizes video content.",
    "callback": "https://yourserver.com/callback",
    "thinking": False,
}

file_path = r"C:\Users\Shivansh Gupta\OneDrive\Desktop\My Projects\AI-Projects\Memories-AI\Videos\Normal\Oct_16__1700_30s_202510171224_v00pk.mp4"


files = [
    ("req", ("req.json", json.dumps(data), "application/json")),
    ("files", (os.path.basename(file_path), open(file_path, 'rb'), "video/mp4")),
]

response = requests.post(url, files=files, headers=headers)
print(response.json())
```
